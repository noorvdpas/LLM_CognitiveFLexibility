```python
import pandas as pd
import statsmodels.formula.api as smf

true_results = pd.read_csv("results_isot_true.csv")
fake_results = pd.read_csv("results_isot_fake.csv")

true_results = true_results[true_results['context_type'].isin(['clean', 'meani'])].copy()
fake_results = fake_results[fake_results['context_type'].isin(['clean', 'meani'])].copy()

true_results['content'] = 'isot_true'
fake_results['content'] = 'isot_false'

data = pd.concat([true_results, fake_results], ignore_index=True)

data['content'] = data['content'].astype('category').cat.set_categories(['isot_true', 'isot_false'], ordered=True)
data['context_type'] = data['context_type'].astype('category').cat.set_categories(['clean', 'meani'], ordered=True)

data_model = data[data['model_name'] == "meta-llama/Llama-3.1-8B"].copy()
ols_model = smf.ols("accuracy ~ content * context_type + scale(text_length)",data=data_model).fit()

print(ols_model.summary())
```
